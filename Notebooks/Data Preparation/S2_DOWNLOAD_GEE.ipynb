```python
import re
import time
import requests
import pandas as pd
from pathlib import Path
from datetime import datetime, timedelta, timezone
from concurrent.futures import ThreadPoolExecutor, as_completed
from pystac_client import Client

# ── Config ─────────────────────────────────────────────────────────────────────
CATALOG_URL      = "https://earth-search.aws.element84.com/v1"
CDSE_CATALOG_URL = "https://catalogue.dataspace.copernicus.eu/stac"
COLLECTION       = "sentinel-2-l2a"
CDSE_COLLECTION  = "SENTINEL-2"
OUTPUT_ROOT      = Path(r'E:\sentinel_lake_ice\Data\Input\Breakup S2 Data')
MAX_WORKERS      = 10

BANDS = ["B01", "B02", "B03", "B04", "B05", "B06",
         "B07", "B08", "B8A", "B09", "B11", "B12", "SCL"]

ASSET_KEY_MAP = {
    "B01": "coastal",  "B02": "blue",      "B03": "green",
    "B04": "red",      "B05": "rededge1",  "B06": "rededge2",
    "B07": "rededge3", "B08": "nir",       "B8A": "nir08",
    "B09": "nir09",    "B11": "swir16",    "B12": "swir22",
    "SCL": "scl",
}

# CDSE uses band names directly as asset keys ("B02", "B8A", etc.)
# rather than earth-search's descriptive names ("blue", "nir08", etc.).
# _find_asset_href() tries both conventions so the same download_scene()
# function works regardless of which catalog resolved the item.
def _find_asset_href(assets: dict, band: str) -> str | None:
    for key in (ASSET_KEY_MAP[band], band, band.lower()):
        if key in assets and "href" in assets[key]:
            return assets[key]["href"]
    return None


# ── ID parsing ─────────────────────────────────────────────────────────────────

def parse_gee_id(gee_id):
    m = re.fullmatch(
        r'(\d{8}T\d{6})_(\d{8}T\d{6})_(T[0-9]{2}[A-Z]{3})',
        gee_id
    )
    if not m:
        raise ValueError(f"Unexpected GEE ID format: {gee_id!r}")
    sensing_str, _, tile_with_t = m.groups()
    sensing_dt = datetime.strptime(sensing_str, "%Y%m%dT%H%M%S").replace(
        tzinfo=timezone.utc)
    mgrs_no_t = tile_with_t[1:]          # e.g. "06WWU"
    # earth-search item IDs drop the leading zero on the zone number:
    # "06WWU" → "6WWU", "04WFD" → "4WFD", "12ABC" stays "12ABC"
    zone_int    = int(mgrs_no_t[:2])
    mgrs_for_id = str(zone_int) + mgrs_no_t[2:]
    return {
        "sensing_dt":  sensing_dt,
        "mgrs_tile":   tile_with_t,
        "date_str":    sensing_str[:8],
        "mgrs_no_t":   mgrs_no_t,         # "06WWU" — property queries & CDSE IDs
        "mgrs_for_id": mgrs_for_id,        # "6WWU"  — earth-search item ID
    }


# ── earth-search helpers ───────────────────────────────────────────────────────

def _fetch_item_direct(stac_id, session):
    url = f"{CATALOG_URL}/collections/{COLLECTION}/items/{stac_id}"
    r = session.get(url, timeout=30)
    return r.json() if r.status_code == 200 else None


def _resolve_via_earthsearch(parts, session, catalog):
    """Try earth-search. Returns (stac_id, item_dict) or (None, None)."""
    mgrs_for_id = parts["mgrs_for_id"]
    mgrs_no_t   = parts["mgrs_no_t"]
    date        = parts["date_str"]

    # Fast direct fetch using known ID patterns
    for stac_id in [
        f"S2A_{mgrs_for_id}_{date}_0_L2A",
        f"S2B_{mgrs_for_id}_{date}_0_L2A",
        f"S2C_{mgrs_for_id}_{date}_0_L2A",
        f"S2A_{mgrs_for_id}_{date}_1_L2A",
        f"S2B_{mgrs_for_id}_{date}_1_L2A",
    ]:
        data = _fetch_item_direct(stac_id, session)
        if data and "assets" in data:
            item_dt_str = data.get("properties", {}).get("datetime", "")
            if item_dt_str:
                item_dt = datetime.fromisoformat(item_dt_str.replace("Z", "+00:00"))
                if abs((item_dt - parts["sensing_dt"]).total_seconds()) <= 90:
                    return stac_id, data
            else:
                return stac_id, data

    # Slower datetime-window + tile property search
    window_start = parts["sensing_dt"] - timedelta(seconds=90)
    window_end   = parts["sensing_dt"] + timedelta(seconds=90)
    dt_str = (
        f"{window_start.strftime('%Y-%m-%dT%H:%M:%SZ')}/"
        f"{window_end.strftime('%Y-%m-%dT%H:%M:%SZ')}"
    )
    try:
        search = catalog.search(
            collections=[COLLECTION],
            datetime=dt_str,
            query={"s2:mgrs_tile": {"eq": mgrs_no_t}},
        )
        items = list(search.items())
        if items:
            return items[0].id, items[0].to_dict()
    except Exception as e:
        print(f"    ⚠  earth-search STAC query error: {e}")

    return None, None


# ── CDSE fallback ──────────────────────────────────────────────────────────────

def _resolve_via_cdse(parts, session):
    """
    Fall back to Copernicus Data Space Ecosystem STAC.

    CDSE holds the full Sentinel-2 archive back to 2015, including
    antimeridian-adjacent zone-06 tiles that earth-search skips.

    CDSE does NOT reliably support the 'query' extension, so we search
    by datetime only and filter locally.  CDSE item IDs always embed the
    tile as '_T06WWU_', so string matching is unambiguous.

    Example CDSE item ID:
        S2A_MSIL2A_20190415T212521_N0207_R071_T06WWU_20190416T025348
    """
    tile_tag     = f"_T{parts['mgrs_no_t']}_"   # e.g. "_T06WWU_"
    mgrs_no_t    = parts["mgrs_no_t"]

    # Use a wider window (±5 min) because we're not filtering in the query
    window_start = parts["sensing_dt"] - timedelta(minutes=5)
    window_end   = parts["sensing_dt"] + timedelta(minutes=5)
    dt_str = (
        f"{window_start.strftime('%Y-%m-%dT%H:%M:%SZ')}/"
        f"{window_end.strftime('%Y-%m-%dT%H:%M:%SZ')}"
    )

    try:
        cdse_catalog = Client.open(CDSE_CATALOG_URL)
    except Exception as e:
        print(f"    ⚠  Could not open CDSE catalog: {e}")
        return None, None

    # Pass 1: try with query extension (faster when supported)
    # Pass 2: plain datetime search + local tile filter (always works)
    for attempt, kwargs in enumerate([
        dict(collections=[CDSE_COLLECTION], datetime=dt_str,
             query={"s2:mgrs_tile": {"eq": mgrs_no_t}}),
        dict(collections=[CDSE_COLLECTION], datetime=dt_str),
    ]):
        try:
            search = cdse_catalog.search(**kwargs)
            items  = list(search.items())

            if attempt == 1:
                # Filter locally: tile tag appears in item ID and/or properties
                items = [
                    i for i in items
                    if tile_tag in i.id
                    or i.properties.get("s2:mgrs_tile") == mgrs_no_t
                ]

            if items:
                # Pick the item whose sensing time is closest to the GEE timestamp
                def _dt_delta(item):
                    dt_str = item.properties.get("datetime", "")
                    if not dt_str:
                        return float("inf")
                    dt = datetime.fromisoformat(dt_str.replace("Z", "+00:00"))
                    return abs((dt - parts["sensing_dt"]).total_seconds())

                best = min(items, key=_dt_delta)
                return best.id, best.to_dict()

        except Exception as e:
            if attempt == 0:
                # query extension not supported — silently retry without it
                continue
            print(f"    ⚠  CDSE search failed: {e}")

    return None, None


# ── Unified resolver ───────────────────────────────────────────────────────────

def resolve_stac_item(gee_id, session, catalog):
    parts = parse_gee_id(gee_id)

    stac_id, item_dict = _resolve_via_earthsearch(parts, session, catalog)
    if stac_id:
        return stac_id, item_dict, "earth-search"

    stac_id, item_dict = _resolve_via_cdse(parts, session)
    if stac_id:
        return stac_id, item_dict, "cdse"

    return None, None, None


# ── Download helpers ───────────────────────────────────────────────────────────

def download_file(url, out_path, session):
    """Download url → out_path, skipping if already present."""
    if out_path.exists():
        return
    r = session.get(url, stream=True, timeout=120)
    r.raise_for_status()
    with open(out_path, "wb") as f:
        for chunk in r.iter_content(65536):
            f.write(chunk)


def download_scene(stac_id, item_dict, site_dir, session):
    scene_dir = site_dir / stac_id
    scene_dir.mkdir(parents=True, exist_ok=True)
    assets  = item_dict.get("assets", {})
    missing = []
    for band in BANDS:
        href = _find_asset_href(assets, band)
        if not href:
            missing.append(band)
            continue
        download_file(href, scene_dir / f"{band}.tif", session)
    return missing


# ── Main ──────────────────────────────────────────────────────────────────────

def main():
    OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
    catalog = Client.open(CATALOG_URL)

    from collections import defaultdict
    by_site = defaultdict(list)
    for site_code, gee_id in INPUT_SCENES:
        by_site[site_code].append(gee_id)

    for site_code, gee_ids in by_site.items():
        print(f"\n{'='*80}")
        print(f"Site: {site_code}  ({len(gee_ids)} scenes)")
        print(f"{'='*80}")

        site_dir = OUTPUT_ROOT / site_code
        site_dir.mkdir(parents=True, exist_ok=True)

        print("  Resolving GEE IDs → STAC IDs ...")
        resolved   = {}
        unresolved = []

        with requests.Session() as session:
            for gee_id in gee_ids:
                stac_id, item_dict, source = resolve_stac_item(gee_id, session, catalog)
                if stac_id:
                    resolved[gee_id] = (stac_id, item_dict, source)
                    print(f"    ✓  {gee_id}  →  {stac_id}  [{source}]")
                else:
                    unresolved.append(gee_id)
                    print(f"    ✗  {gee_id}  →  NOT FOUND (checked earth-search + CDSE)")
                time.sleep(0.1)

        if unresolved:
            print(f"\n  ⚠  {len(unresolved)} scene(s) still not found after both catalogs:")
            for gee_id in unresolved:
                print(f"       {gee_id}")

        if not resolved:
            print("  No scenes resolved — skipping downloads.")
            continue

        print(f"\n  Downloading {len(BANDS)} bands per scene ({MAX_WORKERS} workers)...")
        metadata  = []
        completed = 0

        def _worker(args):
            gee_id, (stac_id, item_dict, source) = args
            with requests.Session() as sess:
                # NOTE: CDSE requires an OAuth2 Bearer token for actual downloads.
                # See: https://documentation.dataspace.copernicus.eu/APIs/Token.html
                # Add:  sess.headers.update({"Authorization": f"Bearer {token}"})
                missing = download_scene(stac_id, item_dict, site_dir, sess)
            props = item_dict.get("properties", {})
            return gee_id, stac_id, props, missing, source

        with ThreadPoolExecutor(max_workers=MAX_WORKERS) as pool:
            futures = {pool.submit(_worker, item): item
                       for item in resolved.items()}
            for future in as_completed(futures):
                gee_id, stac_id, props, missing, source = future.result()
                completed += 1
                status = "✗" if missing else "✓"
                print(f"  [{completed:>4}/{len(resolved)}] {status} {stac_id} [{source}]")
                if missing:
                    print(f"    missing bands: {missing}")
                metadata.append({
                    'gee_id':        gee_id,
                    'stac_id':       stac_id,
                    'source':        source,
                    'datetime':      props.get('datetime'),
                    'cloud_cover':   props.get('eo:cloud_cover', -1),
                    'mgrs_tile':     props.get('s2:mgrs_tile'),
                    'missing_bands': ','.join(missing) if missing else '',
                })

        meta_df   = pd.DataFrame(metadata)
        meta_path = site_dir / 'metadata.csv'
        meta_df.to_csv(meta_path, index=False)

        n_es   = sum(1 for _, _, s in resolved.values() if s == 'earth-search')
        n_cdse = sum(1 for _, _, s in resolved.values() if s == 'cdse')

        print(f"\n  {site_code} complete")
        print(f"    Resolved  : {len(resolved)}/{len(gee_ids)}")
        print(f"      earth-search : {n_es}")
        print(f"      cdse         : {n_cdse}")
        print(f"    Avg cloud : {meta_df['cloud_cover'].mean():.1f}%")
        print(f"    Metadata  : {meta_path}")

    print(f"\n{'='*80}\nALL DONE\n{'='*80}")


if __name__ == "__main__":
    main()
```
